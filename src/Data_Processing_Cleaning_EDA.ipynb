```python
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.ops import nearest_points
from matplotlib_scalebar.scalebar import ScaleBar

# After loading the GeoDataFrames, convert to degree decimal CRS
lineaments_gdf = lineaments_gdf.to_crs("EPSG:4326")
nogold_gdf = nogold_gdf.to_crs("EPSG:4326")

# For distance calculation, temporarily transform to UTM inside the function
def calculate_distance_and_layer(point, lineaments_gdf):
    # Transform geometries to UTM for distance calc
    lineaments_utm = lineaments_gdf.to_crs("EPSG:32643")
    point_utm = gpd.GeoSeries([point], crs="EPSG:4326").to_crs("EPSG:32643").iloc[0]

    lineament_distances = lineaments_utm.geometry.distance(point_utm)
    min_distance_idx = lineament_distances.idxmin()
    nearest_layer = lineaments_gdf.loc[min_distance_idx, 'layer']  # Use original CRS for layer attribute
    return lineament_distances.min(), nearest_layer

# Rest of your code stays the same
results = nogold_gdf.geometry.apply(lambda point: calculate_distance_and_layer(point, lineaments_gdf))
nogold_gdf['dist to fault'] = results.apply(lambda x: x[0])  # Distance in meters
nogold_gdf['lineament'] = results.apply(lambda x: x[1])

# Print the updated GeoDataFrame
print(nogold_gdf[['dist to fault', 'lineament']])  # Adjust columns as needed for your context

import math

# Calculate longitude distance for the map's center latitude
center_latitude = nogold_gdf.geometry.centroid.y.mean()  # Average latitude of the points
longitude_km_per_degree = 111.32 * math.cos(math.radians(center_latitude))  # Longitude in km per degree
latitude_km_per_degree = 111.32  # Latitude in km per degree (constant)

# Calculate 0.25° in kilometers
longitude_km_025 = longitude_km_per_degree * 0.25
latitude_km_025 = latitude_km_per_degree * 0.25

# Plot text reflecting 0.25° equivalence on the map
fig, ax = plt.subplots(figsize=(12, 12))

nogold_plot = nogold_gdf.to_crs("EPSG:4326")
nogold_proj = nogold_gdf.to_crs("EPSG:32643")

# Plot lineaments
unique_layers = lineaments_plot['layer'].unique()
colors = plt.cm.tab10.colors  # Up to 10 distinct colors
for i, layer in enumerate(unique_layers):
    layer_data = lineaments_plot[lineaments_plot['layer'] == layer]
    layer_data.plot(ax=ax, edgecolor=colors[i % len(colors)], linewidth=1.5, label=f"Layer: {layer}")

# Plot NoGold points
nogold_plot.plot(ax=ax, color='gray', markersize=70, marker='o', label='NoGold Points', alpha=0.7, edgecolor='black')

# Annotate NoGold points
for idx, row in nogold_plot.iterrows():
    geom = row.geometry
    dist = nogold_proj.loc[idx, 'dist to fault']  # Distance in meters
    layer = nogold_proj.loc[idx, 'lineament']
    if geom.geom_type == 'Point':
        ax.text(geom.x, geom.y, f"{dist:.2f}m ({layer})", fontsize=9, color='red', ha='right', weight='bold')

# Add text for the 0.25° equivalence on the map
ax.text(0.02, 0.02, f"0.25° ≈ {longitude_km_025:.2f} km (lon), {latitude_km_025:.2f} km (lat)",
        fontsize=12, color='blue', transform=ax.transAxes, ha='left', weight='bold')

# Customize plot
plt.legend(title="Map Layers", fontsize=10, title_fontsize=12)
plt.title("Lineaments and NoGold Points with Distances", fontsize=16)
plt.xlabel("Longitude (degrees)", fontsize=12)
plt.ylabel("Latitude (degrees)", fontsize=12)
plt.grid(True)
plt.axis('equal')
plt.show()

# Save the updated NoGold GeoDataFrame to a new shapefile
output_path = r"D:\Projects\GSI Hackathon_2025\Python Working Files\Updated_NoGold_Points.shp"
nogold_gdf.to_file(output_path)
```
